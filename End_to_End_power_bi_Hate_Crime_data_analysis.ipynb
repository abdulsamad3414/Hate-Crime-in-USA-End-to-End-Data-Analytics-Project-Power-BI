# End to End Data Analysis By Using Python and Power Bi

```python
import pandas as pd
df=pd.read_csv("/content/hate_crime-csv-2-csv.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
# drop these columns
df.drop("ORI",axis=1,inplace=True)
df.drop("STATE_ABBR",axis=1,inplace=True)
df.drop("PUB_AGENCY_UNIT",axis=1,inplace=True)
df.drop("POPULATION_GROUP_CODE",axis=1,inplace=True)
df.drop("POPULATION_GROUP_DESC",axis=1,inplace=True)
df.drop("Bais_Description",axis=1,inplace=True)
df.drop("OFFENDER_ETHNICITY",axis=1,inplace=True)
df.drop("ADULT_VICTIM_COUNT",axis=1,inplace=True)
df.drop("JUVENILE_VICTIM_COUNT",axis=1,inplace=True)
df.drop("ADULT_OFFENDER_COUNT",axis=1,inplace=True)
df.drop("JUVENILE_OFFENDER_COUNT",axis=1,inplace=True)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
# offender race fill with mode
df["OFFENDER_RACE"].fillna(df["OFFENDER_RACE"].mode()[0],inplace=True)
```

```python
# total indivitaul vixtims fill with median
df["TOTAL_INDIVIDUAL_VICTIMS"].fillna(df["TOTAL_INDIVIDUAL_VICTIMS"].median(),inplace=True)
```

```python
df.head()
```

# Data Visualization


```python
import plotly.express as px
import plotly.graph_objects as go
```

# Yearly Hate Crime Trend

```python
yearly = df.groupby('DATA_YEAR')['INCIDENT_ID'].count().reset_index()

fig = px.line(
    yearly,
    x='DATA_YEAR',
    y='INCIDENT_ID',
    markers=True,
    title='Yearly Hate Crime Trend'
)

fig.show()
```

# Top 10 States by Hate Crime Incidents



```python
top_states = (
    df['STATE_NAME']
    .value_counts()
    .head(10)
    .reset_index()
)

top_states.columns = ['State', 'Incidents']

fig = px.bar(
    top_states,
    x='Incidents',
    y='State',
    orientation='h',
    title='Top 10 States by Hate Crime Incidents'
)

fig.show()
```

# Top Bias Motivations

```python
top_bias = (
    df['BIAS_DESC']
    .value_counts()
    .head(10)
    .reset_index()
)

top_bias.columns = ['Bias', 'Incidents']

fig = px.bar(
    top_bias,
    x='Incidents',
    y='Bias',
    orientation='h',
    title='Top Bias Motivations'
)

fig.show()
```

# Offense Type Analysis

```python
offense = (
    df['OFFENSE_NAME']
    .value_counts()
    .head(10)
    .reset_index()
)

offense.columns = ['Offense', 'Count']

fig = px.bar(
    offense,
    x='Count',
    y='Offense',
    orientation='h',
    title='Top Offense Types'
)

fig.show()
```

# Crime Location Analysis

```python
locations = (
    df['LOCATION_NAME']
    .value_counts()
    .head(10)
    .reset_index()
)

locations.columns = ['Location', 'Count']

fig = px.bar(
    locations,
    x='Count',
    y='Location',
    orientation='h',
    title='Top Crime Locations'
)

fig.show()
```

# Offender Race Distribution

```python
race = (
    df['OFFENDER_RACE']
    .value_counts()
    .reset_index()
)

race.columns = ['Race', 'Count']

fig = px.pie(
    race,
    names='Race',
    values='Count',
    title='Offender Race Distribution'
)

fig.show()
```

# Victim Type Analysis

```python
victims = (
    df['VICTIM_TYPES']
    .value_counts()
    .head(10)
    .reset_index()
)

victims.columns = ['Victim Type', 'Count']

fig = px.treemap(
    victims,
    path=['Victim Type'],
    values='Count',
    title='Victim Type Analysis'
)

fig.show()
```

# Multiple Offense Analysis

```python
fig = px.pie(
    df,
    names='MULTIPLE_OFFENSE',
    title='Multiple Offense Analysis'
)

fig.show()
```

# Region Analysis

```python
region = (
    df['REGION_NAME']
    .value_counts()
    .reset_index()
)

region.columns = ['Region', 'Count']

fig = px.bar(
    region,
    x='Region',
    y='Count',
    title='Incidents by Region'
)

fig.show()
```

# Correlation Heatmap

```python
corr = df.select_dtypes(include='number').corr()

fig = px.imshow(
    corr,
    text_auto=True,
    aspect='auto',
    title='Correlation Heatmap'
)

fig.show()
```
